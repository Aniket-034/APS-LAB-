# Shortest Path Using Dijkstra Algorithm in Python
Problem Description:
Given a weighted graph with non negative edge weights. Find the shortest distance from a source vertex to all other vertices using Dijkstra algorithm

#Theory
Dijkstra algorithm finds shortest distance from one source node to all other nodes in a weighted graph with non negative weights. Dictionary structure stores nodes as keys and neighbor nodes with weights as values.

##Algorithm Steps

1. Represent graph using dictionary.
2. Set distance of every node as infinity.
3. Set distance of source node as 0.
4. Select node with smallest distance among unvisited nodes.
5. Update distance of neighbor nodes.
6. Repeat until all nodes get visited.

```python
def dijkstra(graph, start):

    distances = {}
    predecessors = {}
    visited = []

    for node in graph:
        distances[node] = float('inf')
        predecessors[node] = None

    distances[start] = 0

    while len(visited) < len(graph):

        min_node = None
        min_distance = float('inf')

        for node in graph:
            if node not in visited and distances[node] < min_distance:
                min_distance = distances[node]
                min_node = node

        if min_node is None: # Handle disconnected graphs
            break

        visited.append(min_node)

        for neighbor, weight in graph[min_node].items():
            new_distance = distances[min_node] + weight

            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                predecessors[neighbor] = min_node

    return distances, predecessors
```

```python
graph = {
'A': {'B': 4, 'C': 2},
'B': {'A': 4, 'C': 5, 'D': 10},
'C': {'A': 2, 'B': 5, 'D': 3},
'D': {'B': 10, 'C': 3}
}

result = dijkstra(graph, 'A')

for node in result:
    print("Distance from A to", node, "=", result[node])
```

## Reconstruct and Display Shortest Path

```python
def get_shortest_path(predecessors, start, target):
    path = []
    current = target
    while current is not None:
        path.insert(0, current)
        current = predecessors[current]
    if path[0] == start:
        return ' -> '.join(path)
    else:
        return "No path found"

```

## User Input Graph and Testing

```python
def create_graph_from_input():
    graph = {}
    num_nodes = int(input("Enter the number of nodes in the graph: "))
    nodes = [input(f"Enter name for node {i+1}: ") for i in range(num_nodes)]

    for node in nodes:
        graph[node] = {}

    print("\nEnter edges and weights. Enter 'done' when finished.")
    while True:
        edge_input = input("Enter edge (source target weight, e.g., A B 5): ")
        if edge_input.lower() == 'done':
            break
        try:
            source, target, weight_str = edge_input.split()
            weight = int(weight_str)
            if source not in graph: # Automatically add new nodes if mentioned in edge
                graph[source] = {}
            if target not in graph:
                graph[target] = {}
            graph[source][target] = weight
            # For an undirected graph, add the reverse edge too
            # graph[target][source] = weight
        except ValueError:
            print("Invalid input. Please use format 'source target weight'.")
    return graph

# Get graph from user input
print("--- Creating Graph ---")
user_graph = create_graph_from_input()
print("\nYour graph:", user_graph)

# Test with different source nodes
while True:
    start_node = input("\nEnter a source node to calculate shortest paths (or 'done' to finish): ")
    if start_node.lower() == 'done':
        break
    if start_node not in user_graph:
        print("Source node not found in the graph. Please try again.")
        continue

    distances, predecessors = dijkstra(user_graph, start_node)

    print(f"\nShortest distances from {start_node}:")
    for node, dist in distances.items():
        if dist == float('inf'):
            print(f"  To {node}: Unreachable")
        else:
            print(f"  To {node}: {dist}")
            path = get_shortest_path(predecessors, start_node, node)
            print(f"    Path: {path}")
```
